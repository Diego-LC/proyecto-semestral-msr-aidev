# 00 - Setup y Rutas (AIDev)

Este notebook valida el entorno (`.venv`) y define rutas canónicas para ejecutar el pipeline real del proyecto (sampling → cards).

Outputs que se generan/usan en notebooks posteriores:
- `exploration/aidev/sampling/outputs/merged_after_rework_sample.csv`
- `exploration/aidev/sampling/outputs/merged_after_rework_sample_summary.json`
- `exploration/aidev/preparation/outputs/merged_after_rework_cards_seed_20260510.csv`
- `exploration/aidev/preparation/outputs/merged_after_rework_cards_seed_20260510_summary.json`


```python
from __future__ import annotations

import json
import os
import subprocess
from pathlib import Path

PROJECT_ROOT = Path.cwd()
PY = PROJECT_ROOT / '.venv' / 'bin' / 'python'

print(f'[info] PROJECT_ROOT={PROJECT_ROOT}')
if not PY.exists():
    raise FileNotFoundError(
        f'[error] No existe {PY}. Crea el venv con: python3 -m venv .venv'
    )
print(f'[ok] Usando PY={PY}')

def run(cmd: list[str]) -> subprocess.CompletedProcess[str]:
    print('[cmd]', ' '.join(map(str, cmd)))
    return subprocess.run(cmd, check=True, text=True, capture_output=True)

# Verificar dependencias mínimas (pandas + pyarrow).
check = run([str(PY), '-c', 'import pandas, pyarrow; import sys; print(sys.executable)'])
print('[ok] pandas/pyarrow import OK')
print(check.stdout.strip())

# Rutas canónicas (se pueden sobreescribir en notebooks siguientes).
SAMPLE_CSV = PROJECT_ROOT / 'exploration/aidev/sampling/outputs/merged_after_rework_sample.csv'
SAMPLE_SUMMARY = PROJECT_ROOT / 'exploration/aidev/sampling/outputs/merged_after_rework_sample_summary.json'
CARDS_CSV = PROJECT_ROOT / 'exploration/aidev/preparation/outputs/merged_after_rework_cards_seed_20260510.csv'
CARDS_SUMMARY = PROJECT_ROOT / 'exploration/aidev/preparation/outputs/merged_after_rework_cards_seed_20260510_summary.json'
REPORTS_DIR = PROJECT_ROOT / 'exploration/aidev/reports'

print('[info] Rutas definidas:')
print(' -', SAMPLE_CSV)
print(' -', SAMPLE_SUMMARY)
print(' -', CARDS_CSV)
print(' -', CARDS_SUMMARY)
print(' -', REPORTS_DIR)

REPORTS_DIR.mkdir(parents=True, exist_ok=True)
print('[ok] Reports dir listo')
```
